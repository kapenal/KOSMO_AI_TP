```python
### ✅ 셀 1: SBERT 모델 학습 및 저장

# 1. 라이브러리 로드
import json
import joblib
from sentence_transformers import SentenceTransformer
import numpy as np

# 2. 경로 설정
DATA_PATH = "C:/kosmo/data/"
SAVE_PATH = "C:/kosmo/test_data/IH/"
FILE_NAME = "movies_DB_partial.json"

# 3. 데이터 로드
with open(DATA_PATH + FILE_NAME, encoding="utf-8") as f:
    movies = json.load(f)

# 4. 텍스트 결합 함수 (장르 가중치 포함)
def combine_text(overview, genres, genre_weight=3):
    overview = overview or ""
    genres = genres or []
    genres_repeated = " ".join(genres * genre_weight)
    return f"{overview} 장르: {genres_repeated}"

# 5. 전체 문장 만들기
genre_weight = 3
texts = [combine_text(m.get("overview"), m.get("genres"), genre_weight) for m in movies]

# 6. SBERT 모델 로드 및 벡터 생성
model = SentenceTransformer("snunlp/KR-SBERT-V40K-klueNLI-augSTS")
embeddings = model.encode(texts, convert_to_numpy=True , batch_size=64, show_progress_bar=True)

# 7. 저장
joblib.dump(movies, SAVE_PATH + "movies.pkl")
joblib.dump(embeddings, SAVE_PATH + "sbert_vectors.pkl")
model.save(SAVE_PATH + "sbert_model")
print("✅ 모델 학습 및 저장 완료")
```
